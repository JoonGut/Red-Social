```python
import pandas as pd
import numpy as np
df = pd.read_csv('reto.csv')
df.head()
```

cargamos el dataset, que contiene información sobre publicaciones, usuarios e interacciones en una red social.

```python
df.info()
```

Aqui vemos cuántas filas y columnas hay, que nos pondra que tenemos en el dataset 12.000 registros y 28 variables relacionadas con usuarios, publicaciones e interacción.

```python
df.describe()
```

La parte esta nos da un resumen estadistico de las columnas numericas

```python
df.isnull().sum()
```

Aqui detectamos los valores nulos en la columna mentions, lo cual es habitual cuando una publicación no menciona a ningún usuario.

```python
df = df.drop_duplicates()
```

Esta línea eliminamos las filas duplicadas del DataFrame, garantizando que cada publicación se cuente una sola vez. Esto asegura que los análisis y estadísticas posteriores sean precisos y evita registros repetidos

```python
df['mentions'] = df['mentions'].fillna('None')
```

Esta línea reemplaza los valores faltantes en la columna mentions por 'None', asegurando que todas las publicaciones tengan un valor definido. Esto permite analizar correctamente la presencia de menciones y evita problemas en cálculos o gráficos posteriores.

```python
df['timestamp'] = pd.to_datetime(df['timestamp'])
```

Convertimos la columna timestamp al tipo de dato datetime, permitiendo analizar correctamente la actividad de los usuarios a lo largo del tiempo y crear agrupaciones temporales precisas, como por mes, día o hora.

```python
df['total_interactions'] = df['likes_count'] + df['comments_count'] + df['shares_count']
```

Calculamos el total de interacciones de cada publicación sumando “me gusta”, comentarios y compartidos. Permite analizar el engagement general y sirve como métrica central para comparar publicaciones, campañas o tipos de contenido.

```python
df['month'] = df['timestamp'].dt.month
```

Extraemos el número de mes de la columna de fecha, permitiendo analizar la actividad de los usuarios y las interacciones de manera mensual. Esto facilita identificar tendencias, patrones estacionales y meses con mayor o menor participación.

```python
q99 = df['total_interactions'].quantile(0.99)
df = df[df['total_interactions'] <= q99]
```

Filtramos el outliers

Tras este proceso de limpieza y preprocesamiento, los datos ya estárian preparados para realizar un análisis exploratorio y extraer conclusiones relevantes para la toma de decisiones.

```python
df['country'] = df['location']
```

Creamos la columna country a partir de location para poder analizar los datos geográficos y generar una tabla de publicaciones por ubicación de manera más clara

```python
countries = ['Spain', 'USA', 'France', 'Germany', 'Italy']
df['country'] = np.random.choice(countries, size=len(df))
```

asignamos de manera aleatoria un país a cada publicación para poder realizar análisis geográficos y generar tablas o gráficos por ubicación, simulando datos de localización cuando no se cuenta con información real.
